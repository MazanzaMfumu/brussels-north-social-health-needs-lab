```python
# Import 

from pathlib import Path

import geopandas as gpd
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Définir les chemins

ROOT = Path.cwd()

if ROOT.name == "notebooks":
    ROOT = ROOT.parent

PROCESSED_DIR = ROOT / "data" / "processed"
FIGURES_DIR = ROOT / "outputs" / "figures"
MAPS_DIR = ROOT / "outputs" / "maps"

FIGURES_DIR.mkdir(
    parents=True,
    exist_ok=True
)

MAPS_DIR.mkdir(
    parents=True,
    exist_ok=True
)

print("Project root:", ROOT)
print("Figures directory:", FIGURES_DIR)
print("Maps directory:", MAPS_DIR)
```

```python
# Chargement d'un nouveau GeoJSON

file_path = (
    PROCESSED_DIR
    / "brussels_monitoring_vulnerability.geojson"
)

gdf = gpd.read_file(
    file_path
)

print(
    "Shape:",
    gdf.shape
)

print(
    "CRS:",
    gdf.crs
)

gdf.head()
```

```python
# Carte de contrôle des quartiers du Monitoring

ax = gdf.plot(
    figsize=(10, 10),
    edgecolor="black",
    linewidth=0.3
)

ax.set_title(
    "Monitoring neighbourhoods – Brussels-Capital Region"
)

ax.axis("off")

plt.tight_layout()

plt.savefig(
    MAPS_DIR
    / "monitoring_neighbourhoods_control_map.png",
    dpi=200,
    bbox_inches="tight"
)

plt.show()
```

```python
# Première carte analytique : chômage (2023)

ax = gdf.plot(
    column="unemployment_rate",
    legend=True,
    figsize=(10, 10),
    edgecolor="black",
    linewidth=0.2
)

ax.set_title(
    "Unemployment rate by Monitoring neighbourhood – 2023"
)

ax.axis("off")

plt.tight_layout()

plt.savefig(
    MAPS_DIR
    / "monitoring_unemployment_2023_map.png",
    dpi=200,
    bbox_inches="tight"
)

plt.show()
```

```python
# Première carte analytique : cpas (2023)

ax = gdf.plot(
    column="cpas_rate",
    legend=True,
    figsize=(10, 10),
    edgecolor="black",
    linewidth=0.2
)

ax.set_title(
    "CPAS/RIS beneficiary rate by Monitoring neighbourhood – 2023"
)

ax.axis("off")

plt.tight_layout()

plt.savefig(
    MAPS_DIR
    / "monitoring_cpas_2023_map.png",
    dpi=200,
    bbox_inches="tight"
)

plt.show()
```

```python
# # Première carte analytique : bim (2024)

ax = gdf.plot(
    column="bim_rate",
    legend=True,
    figsize=(10, 10),
    edgecolor="black",
    linewidth=0.2
)

ax.set_title(
    "BIM beneficiary rate by Monitoring neighbourhood – 2024"
)

ax.axis("off")

plt.tight_layout()

plt.savefig(
    MAPS_DIR
    / "monitoring_bim_2024_map.png",
    dpi=200,
    bbox_inches="tight"
)

plt.show()
```

## Validation du périmètre de Bruxelles-Nord

Le périmètre d'analyse de Bruxelles-Nord est dérivé de la géographie validée
pour le suivi bruxellois et est mis en correspondance avec la définition
territoriale officielle de Bruxelles-Nord.

Les territoires de suivi ne sont pas considérés comme équivalents aux
quartiers sociosanitaires officiels.

Les territoires sont donc classés au moyen d'une table de correspondance explicite et d'une validation cartographique avant l'application de tout filtre analytique.

```python
from pathlib import Path
import pandas as pd

ROOT = (
    Path.cwd().parent
    if Path.cwd().name == "notebooks"
    else Path.cwd()
)

crosswalk_path = (
    ROOT
    / "data"
    / "processed"
    / "brussels_north_monitoring_crosswalk.csv"
)

if not crosswalk_path.exists():
    raise FileNotFoundError(
        f"Crosswalk not found: {crosswalk_path}"
    )

crosswalk = pd.read_csv(crosswalk_path)

print("Crosswalk rows:", len(crosswalk))
print()

print("Validation status:")
print(
    crosswalk["in_bassin_nord"]
    .value_counts(dropna=False)
)

crosswalk.head(20)
```
