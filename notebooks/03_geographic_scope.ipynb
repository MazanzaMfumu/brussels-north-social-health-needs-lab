```python
# Import 

from pathlib import Path

import geopandas as gpd
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Définir les chemins

ROOT = Path.cwd()

if ROOT.name == "notebooks":
    ROOT = ROOT.parent

PROCESSED_DIR = ROOT / "data" / "processed"
FIGURES_DIR = ROOT / "outputs" / "figures"
MAPS_DIR = ROOT / "outputs" / "maps"

FIGURES_DIR.mkdir(
    parents=True,
    exist_ok=True
)

MAPS_DIR.mkdir(
    parents=True,
    exist_ok=True
)

print("Project root:", ROOT)
print("Figures directory:", FIGURES_DIR)
print("Maps directory:", MAPS_DIR)
```

```python
# Chargement d'un nouveau GeoJSON

file_path = (
    PROCESSED_DIR
    / "brussels_monitoring_vulnerability.geojson"
)

gdf = gpd.read_file(
    file_path
)

print(
    "Shape:",
    gdf.shape
)

print(
    "CRS:",
    gdf.crs
)

gdf.head()
```

```python
# Carte de contrôle des quartiers du Monitoring

ax = gdf.plot(
    figsize=(10, 10),
    edgecolor="black",
    linewidth=0.3
)

ax.set_title(
    "Monitoring neighbourhoods – Brussels-Capital Region"
)

ax.axis("off")

plt.tight_layout()

plt.savefig(
    MAPS_DIR
    / "monitoring_neighbourhoods_control_map.png",
    dpi=200,
    bbox_inches="tight"
)

plt.show()
```

```python
# Première carte analytique : chômage (2023)

ax = gdf.plot(
    column="unemployment_rate",
    legend=True,
    figsize=(10, 10),
    edgecolor="black",
    linewidth=0.2
)

ax.set_title(
    "Unemployment rate by Monitoring neighbourhood – 2023"
)

ax.axis("off")

plt.tight_layout()

plt.savefig(
    MAPS_DIR
    / "monitoring_unemployment_2023_map.png",
    dpi=200,
    bbox_inches="tight"
)

plt.show()
```

```python
# Première carte analytique : cpas (2023)

ax = gdf.plot(
    column="cpas_rate",
    legend=True,
    figsize=(10, 10),
    edgecolor="black",
    linewidth=0.2
)

ax.set_title(
    "CPAS/RIS beneficiary rate by Monitoring neighbourhood – 2023"
)

ax.axis("off")

plt.tight_layout()

plt.savefig(
    MAPS_DIR
    / "monitoring_cpas_2023_map.png",
    dpi=200,
    bbox_inches="tight"
)

plt.show()
```
